Time series analysis: plot monthly and quarterly sales trends using line charts

```python
## Convert Date from text to datetime
df["Date"] = pd.to_datetime(df["Date"])

## Sort by date
df = df.sort_values("Date")
```

```python
df.head()
```

```python
monthly_sales = (
    df.set_index("Date")["Total Amount"]
      .resample("MS")
      .sum()
)

monthly_sales
```

```python
quarterly_sales = (
    df.set_index("Date")["Total Amount"]
      .resample("QS")
      .sum()
)

quarterly_sales
```

```python
fig, axes = plt.subplots(2, 1, figsize=(12, 10))

monthly_sales.plot(
    ax=axes[0],
    kind="line",
    marker="o",
    color="steelblue"
)
axes[0].set_title("Monthly Sales Trend")
axes[0].set_xlabel("Month")
axes[0].set_ylabel("Total Sales")
axes[0].grid(True)

quarterly_sales.plot(
    ax=axes[1],
    kind="line",
    marker="o",
    color="darkorange"
)
axes[1].set_title("Quarterly Sales Trend")
axes[1].set_xlabel("Quarter")
axes[1].set_ylabel("Total Sales")
axes[1].grid(True)

plt.tight_layout()
plt.show()
```
